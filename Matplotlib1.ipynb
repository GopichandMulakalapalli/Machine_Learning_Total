```python
data={
    "Salary":[25000,30000,37000,28000,39000,48000,55000,52000,35000,25000,34000,98000,60000,56000,45000,12000,65000,71000,75000,99500]
}
```

```python
df.head(5)
```

```python
df.shape
```

```python
# Line Plot:
plt.plot(df["Salary"],marker="*",color="red",linestyle="--")
plt.grid()
plt.show()
```

```python
# Histogram :
plt.hist(df["Salary"],bins=5,color="red")

plt.show()
```

```python
# BoxPlot:
plt.boxplot(df["Salary"])
plt.show()
```

```python
# Univariate : Categorical
```

```python
df["dept"]=["HR","IT","Finance","HR","Finance","IT","HR","Finance","IT","HR"] * 2
```

```python
df.head()
```

```python
df.info()
```

```python
#Pie Chart :
count=df["dept"].value_counts()
count
```

```python
plt.pie(count,labels=count.index,autopct="%1.2f")
plt.axis("equal")
plt.show()
```

```python
#CountPlot
plt.bar(count.index,count,color=["green","black","gold","red"])
plt.show()
```

```python
# Bivariate -- Numerical -- Numerical
df.head()
```

```python
df["Age"]=[22,25,29,24,30,35,40,36,28,22,23,25,26,24,31,29,27,33,35,38]
```

```python
df.head()
```

```python
# Scatter Plot :
plt.scatter(df["Age"],df["Salary"],color="orange")
plt.show()
```

```python
sort_age=df.sort_values("Age")
```

```python
# Line Plot:
plt.plot(sort_age["Age"],df["Salary"],marker="*",color="red",linewidth="2")
plt.grid()
plt.show()
```

```python
# Bar Chart

plt.bar(sort_age["Age"],df["Salary"],color="green")
plt.show()
```

```python
# Bivariante -- Numerical -- Categorical
hr_sal=df[df["dept"]=="HR"]["Salary"]
hr_sal
```

```python
it_sal=df[df["dept"]=="IT"]["Salary"]
it_sal
```

```python
fin_sal=df[df["dept"]=="Finance"]["Salary"]
fin_sal
```

```python
# Boxplot:
plt.boxplot([hr_sal,it_sal,fin_sal],label=["HR","IT","Finance"])
plt.legend()
plt.show()
```

```python
#Pie Chart:
salary_by_dept=df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept,labels=salary_by_dept.index,autopct="%1.2f",shadow='True')
plt.axis("equal")
plt.show()
```

```python
# Calculate average salary
hr_mean = sum(hr_sal) / len(hr_sal)
it_mean = sum(it_sal) / len(it_sal)
fin_mean = sum(fin_sal) / len(fin_sal)
```

```python
plt.bar(["HR","IT","Finance"],[hr_mean,it_mean,fin_mean],color=["green","black","purple"])
plt.grid()
plt.show()
```

```python
# Multivariate Analysis: 3 Numrical
df.head()
```

```python
df["Experience"]=[1,2,4,1.5,5,8,12,10,4,1,1.2,2,3,1.8,5.5,4.5,3.2,7,8,10]
```

```python
df.head()
```

```python
#Bubble Plot:
plt.scatter(df["Age"],df["Salary"],s=df["Experience"]*20,color='Blue',edgecolor='red')
plt.grid()
plt.title("Age Vs Salary Vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# 2 Numerical and 1 Categorical Column:
plt.scatter(df["Age"],df["Salary"],c=df["dept"].map({"HR":"yellow","IT":"blue","Finance":"orange"}))
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age Vs Salary Vs Dept")
plt.legend()
plt.show()
```

```python
color={"HR":"yellow","IT":"blue","Finance":"orange"}

for dept,color in color.items():
  df_dept=df[df["dept"]==dept]
  plt.scatter(df_dept["Age"],df_dept["Salary"],c=color,label=dept)
plt.legend()
```

```python
#Object Oriented API
```

```python
#Line Plot
fig, axs = plt.subplots(1,3,figsize = (15,5))
axs[0].plot(sort_age["Age"],df["Salary"],marker="*",color="red",linewidth="2")
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#Histogram:

axs[1].hist(df["Salary"],bins=5,color="skyblue")
axs[1].grid()
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

#BoxPlot:
axs[2].boxplot(df["Salary"])
axs[2].set_title("BoxPlot")
axs[2].set_xlabel("Salary")
axs[2].set_ylabel("Frequency")

plt.savefig("multipleplots.png")
plt.show()
```

```python
data={
    'Year':[2020,2021,2022,2023],
    'Sales':[100,150,200,250],
    'Profit':[20,30,40,50],
    'Expense':[80,120,160,200]
}
```

```python
df2=pd.DataFrame(data)
```

```python
plt.plot(df2["Year"],df2["Sales"],label="Sales")
plt.plot(df2["Year"],df2["Profit"],label="Profit")
plt.plot(df2["Year"],df2["Expense"],label="Expense")

plt.title("Financial Analysis")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
plt.show()
```
